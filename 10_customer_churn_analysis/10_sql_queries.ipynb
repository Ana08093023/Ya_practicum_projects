## Задание 2 <a id="task2"></a>  
[К содержанию](#soder)  

* Для каждой книги посчитайте количество обзоров и среднюю оценку;

```python
query2 = ''' 

SELECT 
    books.title AS book_name,
    books.book_id AS id,
    SubRev.n_rev AS reviews,
    SubRat.avg_rat AS avg_rating
FROM books LEFT JOIN 
                (SELECT
                    book_id,
                    AVG(rating) AS avg_rat
                FROM
                    ratings
                GROUP BY 
                    book_id
                ) AS SubRat ON SubRat.book_id = books.book_id
        LEFT JOIN
                (SELECT 
                    book_id,
                    COUNT(review_id) AS n_rev
                FROM
                    reviews
                GROUP BY 
                    book_id
                ) AS SubRev ON SubRev.book_id = books.book_id
                
; ''' 

request2 = pd.io.sql.read_sql(query2, con = engine)
request2
```

- Уникальный числовой идентификатор книги из таблицы books.

**Количество обзоров (reviews):**

- Число обзоров, оставленных пользователями для конкретной книги (данные из таблицы reviews).
- Если книга не имеет обзоров, значение равно NULL или 0.
- Пример: *2 обзора для книги "The Body in the Library (Miss Marple #3)".

**Средняя оценка (avg_rating):**

- Средняя оценка книги, рассчитанная на основе данных из таблицы ratings.
- Если у книги нет оценок, значение равно NULL.
- Пример: Средняя оценка для книги "The Body in the Library (Miss Marple #3)" составляет 4.5.

* Определите издательство, которое выпустило наибольшее число книг толще 50 страниц — так вы исключите из анализа брошюры;

```python
query3 = ''' 
SELECT
    T.publisher_id,
    T.publisher,
    T.n_books
FROM
    (SELECT 
        books.publisher_id AS publisher_id,
        COUNT(books.book_id) AS n_books,
        publishers.publisher AS publisher 
    FROM 
        books 
        LEFT JOIN publishers ON books.publisher_id=publishers.publisher_id

    WHERE 
        num_pages>50
    GROUP BY
        books.publisher_id,
        publishers.publisher
    ORDER BY 
        n_books DESC) AS T
LIMIT 1
; ''' 

request3 = pd.io.sql.read_sql(query3, con = engine)
request3
```

## Задание 4 <a id="task4"></a>  
[К содержанию](#soder)  

* Определите автора с самой высокой средней оценкой книг — учитывайте только книги с 50 и более оценками;

```python
query4 = ''' 
SELECT 
    aut_avg.author,
    AVG(aut_avg.avg_rating) AS avg_avg_rating
FROM
        (SELECT 
            authors.author AS author
           , AVG(ratings.rating) AS avg_rating
        FROM books
            LEFT JOIN authors ON books.author_id = authors.author_id
            LEFT JOIN ratings ON books.book_id = ratings.book_id
        GROUP BY 
             books.book_id
             , authors.author_id
        HAVING
            COUNT(ratings.rating_id) >=50 
        ) AS aut_avg 
GROUP BY    
    aut_avg.author
ORDER BY
    avg_avg_rating DESC
LIMIT 5
    
; ''' 

request4 = pd.io.sql.read_sql(query4, con = engine)
request4
```

***Самую высокую оценку получает - J.K. Rowling/Mary GrandPré - 4.283844 (Гарри Поттер ФОРЕВЕР)***

### Вариант 2

```python
query4 = ''' 
SELECT 
    authors.author AS author,
    AVG(sub.avg_rating) AS avg_author_rating
FROM (
        SELECT 
            books.author_id AS author_id,
            AVG(ratings.rating) AS avg_rating
        FROM books
            LEFT JOIN ratings ON books.book_id = ratings.book_id
        GROUP BY books.book_id
        HAVING COUNT(ratings.rating_id) >= 50
    ) AS sub
LEFT JOIN authors ON sub.author_id = authors.author_id
GROUP BY authors.author
ORDER BY avg_author_rating DESC
LIMIT 5;
''' 

request4 = pd.io.sql.read_sql(query4, con=engine)
print(request4)

```

 * Посчитайте среднее количество обзоров от пользователей, которые поставили больше 48 оценок.

***Из предоставленных данных непонятно (неочевидно) являются ли пользователи из таблиц `ratings` и `reviews` одними и теми же, поэтому надо это проверить.***

```python
print('Кол-во пользователей в "reviews" =', reviews['username'].nunique())
print('Кол-во пользователей в "ratings" =', ratings['username'].nunique())
```

**По кол-ву совпадает, но проверим еще по множествам**

```python
set(reviews['username'].unique().tolist()) == set(ratings['username'].unique().tolist())
```

***Множества идентичны. Всё в порядке, это одни и теже пользователи. Можно составлять запрос.***

```python
query5 = ''' 

SELECT
    AVG(Sub.n_reviews)
FROM
    (SELECT
        username,
        COUNT(review_id) AS n_reviews
    FROM
        reviews
    WHERE 
        username IN (
                SELECT 
                    username
                FROM 
                    ratings
                GROUP BY 
                    username
                HAVING 
                    COUNT(rating_id) > 48
                    )
    GROUP BY 
        username
    ) AS Sub
; ''' 

request5 = pd.io.sql.read_sql(query5, con = engine)
request5
```

***Итак среднее количество обзоров от пользователей, которые поставили больше 48 оценок равно 24***

*Проверим*

```python
users = ratings.groupby('username')['rating_id'].count().reset_index()
users = users.query('rating_id > 48')
users
```

```python
users_lst = users["username"].tolist()
# reviews.query('username.isin(@users_lst)')
reviews.query('username in @users_lst').groupby('username')['review_id'].count().mean()
```
